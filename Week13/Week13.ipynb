```python
# NAME: journal.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# create function write_entry
# function should accept file name and a string
# must open in append mode
# each new entry must start on a new line
# close file when finished
# create func read_journal
# accepts file name as parameter
# opens file in read mode
# read entire file and print contents for user
def write_entry(file_name, entry):
    with open(file_name, 'a') as f:
        f.write(entry + '\n')

# PROGRAM LOGIC
# ask user if they want to (W)rite or (R)ead
# if (W) is chosen ask for thoughts and call write_entry
# if (R) call read_journal
# put into while loops so user can keep going until 'quit' is typed
while True:
    action = input('Do you want to (W)rite or (R)ead or (Q)uit? ')
    if action.lower() == 'w':
        entry = input('What do you want to write? ')
        write_entry('journal.txt', entry)
    elif action.lower() == 'r':
        with open('journal.txt') as f:
            for line in f:
                print(line)
    elif action.lower() == 'q':
        break
    else:
        print('Invalid input')
```
