```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

# 1. Define is_palindrome
def is_palindrome(word):
    if word.lower() == word[::-1].lower():
        return True
    else:
        return False
#   -accepts one parameter
#   - Must return true if word is a palindrome and false if not
#   - Check must work regardless of capitalization
# 2. Use string slicing to reverse word
# 3. Prompt User for word
#   -call function using user input
#   - Use returned value to print friendly message indicating result.


# PROGRAM LOGIC
# Call is_palindrome
word = input("Enter a word: ")
if is_palindrome(word):
    print(f"{word} is a palindrome!")
else:
    print(f"{word} is not a palindrome.")
```
