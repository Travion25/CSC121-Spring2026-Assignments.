```python
# NAME: gradebook.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# Function accepts dictionary
# use while loop for student name and grade (0-100) ends when "done" is typed
# student name is key while grade is value
# fruitful function that accepts gradebook dictionary
# calculate and return average grade of all students
gradebook = {}
def add_students(gradebook):
  while True:
    student_name = input("Enter student name (type 'done' to stop): ")
    if student_name.lower() == "done":
      break
    student_grade = int(input("Enter student grade (0-100): "))
    gradebook[student_name] = student_grade
  return gradebook
def calculate_average(gradebook):
  if len(gradebook) == 0:
    return 0
  total_grade = 0
  for student_name, student_grade in gradebook.items():
    total_grade += student_grade
  average_grade = total_grade / len(gradebook)
  return average_grade
# PROGRAM LOGIC
# start empty dictionary gradebook
# call add_students to add to dictionary
# call calculate_average to display class avearage
# print final report showing each student name and grade
print(add_students(gradebook))
if len(gradebook) > 0:
  print(f"Class average: {calculate_average(gradebook):.2f}")
  for student_name, student_grade in gradebook.items():
    print(f"{student_name}: {student_grade}")
```
