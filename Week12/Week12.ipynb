```python
# NAME: rectangle.py

# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# create calculate_rectangle that should be calc area (L*W) and perimeter (2*(l+w))
# return both as a single tuple like area, perimeter
def calculate_rectangle(length, width):
    area = length * width
    perimeter = 2 * (length + width)
    return area, perimeter


# PROGRAM LOGIC
# ask user for for a length and width (convert to floats)
# call calculate_rectangle using user input
calculate_rectangle(5,10)
# unpack return tuple into seperate varibles (my_area and my_perimeter)
# print results for clearly for user
print(calculate_rectangle(5,10))

```
