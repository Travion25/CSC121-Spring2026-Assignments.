```python

# NAME: roster.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

# make add_player
def add_players(roster):
    while True:
      user_option = input("Enter a name (or 'quit' to quit): ")
      if user_option.lower() == "quit":
        break
      else:
        roster.append(user_option)
# use while loop that end when quit is typed
# every name except quit should be added to the list using .append
# fruitful function (is_on_roster) accepts the list and specific name to search for
# function returns true if name is in list false if not
def is_on_roster(roster, name):
  if name.lower() in [player.lower() for player in roster]:
    return True
  else:
    return False
# PROGRAM LOGIC
# create empty list called roster
roster = []
# call add_players to fill list
add_players(roster)
# print roster
print("Roster:")
for player in roster:
  print(player)
# ask user for one specific name to search for
name = input("Enter a name to search for: ")
# call is_on_roster and use returned value to print a message
if is_on_roster(roster, name):
  print(f"{name} is on the roster.")
else:
  print(f"{name} is not on the roster.")
```
